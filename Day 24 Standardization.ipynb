```python
import numpy as np 
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df.head()
```

```python
df = df.iloc[:,2:]
# this code removes the first two columns which are not required for analysis.
```

```python
df.sample(5)
```

```python
# Train Test Split 
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop('Purchased',axis=1),df['Purchased'],test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
############# Standard Scaler #############


from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)
```

```python
# transform train and test sets
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
X_train
```

```python
## standardization
X_train.mean(axis=0)

X_train.std(axis=0)


############# Logistic Regression #############
from sklearn.linear_model import LogisticRegression
logmodel = LogisticRegression()
logmodel.fit(X_train,y_train)
```

```python

# Predictions
predictions = logmodel.predict(X_test)
predictions
```

```python
## plot the predictions vs actual

plt.scatter(y_test,predictions)
```
