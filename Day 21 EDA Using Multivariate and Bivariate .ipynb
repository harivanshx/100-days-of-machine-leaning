```python
import pandas as pd 
import numpy as np
import seaborn as sns

```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
titanic = pd.read_csv('train.csv')

titanic.head()
```

```python
flights = sns.load_dataset('flights')
flights.head()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
# bi variate analysis


#### Scatter Plot  (Numerical vs Numerical  )

sns.scatterplot(x='total_bill',y='tip',data=tips)
```

```python
sns.scatterplot(x='total_bill',y='tip',hue='smoker',data=tips)
```

```python
sns.scatterplot(x='total_bill',y='tip',hue='sex',style="smoker",data=tips,size='size')
```

```python
############### BAR Plot (Categorical vs Numerical) ###############


titanic.head()
```

```python
## average age of passengers per class

sns.barplot(x='Pclass',y='Age',hue='Sex',data=titanic)
```

```python
sns.barplot(x='Pclass',y='Fare',data=titanic)
```

```python
sns.barplot(x='Pclass',y='Fare',hue='Sex',data=titanic)
```

```python
sns.barplot(x='Fare',y='Sex' ,data=titanic)
```

```python
sns.barplot(x='Pclass',y='Fare',hue='Sex',data=titanic)
```

```python
############ Box Plot ###############
sns.boxplot(x='Sex',y='Age',data=titanic, hue='Survived') 
```

```python
################## Distribution Plot (Numerical) ####################


sns.displot(titanic, x='Age', fill=True)
```

```python
## i want both of the distributions on the same plot with kde curves and bars filled differently based on survived or not
# with hist = True
sns.displot(titanic, x='Age', hue='Survived', kind='kde', fill=True)
```

```python
############## Heatmap (Categorical vs Categorical) ###############


titanic.head(3)
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Pclass')['Survived'].mean()*100).plot(kind='bar')
```

```python
(titanic.groupby('Sex')['Survived'].mean()*100).plot(kind='bar')
```

```python

(titanic.groupby('Embarked')['Survived'].mean()*100).plot(kind='bar')
```

```python
# embarked = c and with respect to survived percentage for gender

(titanic.groupby(['Embarked','Sex','Pclass'])['Survived'].mean()*100).unstack().plot(kind='bar')
```

```python
sns.clustermap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
sns.pairplot(titanic, hue='Survived')
```

```python
##### Line Plot ###### ## Numerical vs Numerical ###

sns.lineplot(x='total_bill',y='tip',data=tips)
```

```python
# group by year to get total passengers per month in a year

new = flights.groupby('year')['passengers'].sum()

```

```python
sns.lineplot(x=new.index,y=new.values)
```

```python
x = flights.pivot_table(index='month',columns='year',values='passengers')
```

```python
sns.heatmap(x)
```

```python
sns.clustermap(x)
```
