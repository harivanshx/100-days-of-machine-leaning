```python
##### Constructors in Python
class Person:
    name = "Harry"
    def __init__(self,fullname,marks):
        print(self)
        self.name = fullname 
        self.marks = marks
        print("This is a constructor creating a person object")
p1 = Person("Harivansh", 85)
  
print(p1.name)
print(p1.marks)
print(p1)
```

```python
#### normal constructor
class Animal:
    def __init__(self, species, age):
        self.species = species
        self.age = age
dog = Animal("Dog", 5)
print(dog.species)
print(dog.age)
```

```python
# default constructor
class Cat:
    def __init__(self):
        self.species = "Cat"
        self.age = 3
cat1 = Cat()
print(cat1.species)
print(cat1.age)
# parameterized constructor
class Bird:
    def __init__(self, species, age):
        self.species = species
        self.age = age
bird1 = Bird("Parrot", 2)
print(bird1.species)
print(bird1.age)
```

```python
# mETHODS IN PYTHON OOP



# methods are functions defined inside a class that operate on instances of that class. They can manipulate object state and perform operations related to the object.

class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"

    def get_age(self):
        return f"THe age of dog is {self.age} years! "

    def set_age(self, new_age):
        self.age = new_age
dog1 = Dog("Buddy", 3)
print(dog1.bark())          # Output: Buddy says Woof!


dog1.set_age(44)
print(dog1.get_age())  
```

```python
# create student class that take name and marks of three subjects as arguments in constructor Then create a method to print the average 


class Student:
    def __init__(self, name, mark1, mark2, mark3):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3
        
    def average(self):
        avg = (self.mark1 + self.mark2 + self.mark3) / 3
        return f"The average marks of {self.name} is {avg}"
    
    
    
s1 = Student("Harivansh", 85, 90, 78)
print(s1.average())
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
       
        
    def average(self):
        sum = 0
        for val in self.marks:
            sum += val
        avg = sum / len(self.marks)
        return f"The average marks of {self.name} is {avg}"
    
    
    
s1 = Student("Harivansh",[ 85, 90, 78])
print(s1.average())
```

```python
class Car:
    def __init__(self):
        self.acc = False
        self.brke = False
        self.clutch = False
        
    def start(self):
        self.acc = True
        self.clutch = True
        print("Car has started")
        
        
car1 = Car()
car1.start()
print(car1.acc)
print(car1.brke)
print(car1.clutch)
```

```python
class Account:
    def __init__(self,bal,acc):
        
        self.balance = bal
        self.acc_number = acc
        
    def debit(self,amount):
        self.balance -= amount
        print(f"Debited {amount}. New balance is {self.balance}")
    
    
    def  credit(self,amount):
        self.balance += amount
        print(f"Credited {amount}. New balance is {self.balance}")
        
        
        
    def get_balance(self):
        return self.balance
    
    
    
        
        
        
acc1 = Account(5000, "1234567890")

print(acc1.balance)
print(acc1.acc_number)

acc1.credit(2000)
acc1.debit(1500)


print(acc1.get_balance())

        
```
