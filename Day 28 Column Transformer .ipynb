```python
import pandas as pd 
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder

from sklearn.model_selection import train_test_split

 
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/refs/heads/main/day28-column-transformer/covid_toy.csv")
```

```python
X_train,X_test , y_train, y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
# gender and city are categorical data so you have to use one hot encoding

# cough is oridinal data so what you have to do is use ordinal encoding
 
# in fever we do have 10 column has missing values 
# so in fever we will apply  SimpleImputer


#if you do have label then you have to use label encoder

```

```python
# AAM ZINDIGI

si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[['fever']])

X_test_fever = si.fit_transform(X_test[['fever']])


X_train_fever.shape
```

```python
X_test_fever.shape
```

```python
# ordinalEncoding = >> cough

oe = OrdinalEncoder(categories=[['Mild','Strong']])

X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.fit_transform(X_test[['cough']])


X_test_cough.shape
```

```python
X_train_cough.shape
```

```python
#### one hot encoding on gender and city 


ohe = OneHotEncoder(drop='first')
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city = ohe.fit_transform(X_test[['gender','city']])

X_train_gender_city.shape
```

```python
# Extracction age
X_train_Concat=  np.concatenate((X_train_gender_city,X_train_cough,X_train_fever),axis=1)
X_test_Concat = np.concatenate(X_test_gender_city,X_test_fever,X_test_cough)
X_train_Concat.shape
```
