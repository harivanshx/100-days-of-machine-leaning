```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
cgpa = np.array([6.0, 6.5, 7.0, 7.5, 8.0, 8.5, 9.0, 9.5])
package = np.array([3, 4, 4.5, 5, 6, 7.5, 9, 12])  # in LPA or INR (Lakhs per annum)

df = pd.DataFrame({"CGPA": cgpa, "Package (LPA)": package})
print(df)
```

```python
X = cgpa.reshape(-1, 1)      # independent variable
y = package                  # dependent variable
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])
```

```python
cgpa_input = 8.7
prediction = model.predict([[cgpa_input]])
print(f"Predicted Package for CGPA {cgpa_input}: {prediction[0]:.2f} LPA")

```

```python
plt.scatter(cgpa, package, label="Actual Data")
plt.plot(cgpa, model.predict(X), label="Best Fit Line")
plt.xlabel("CGPA")
plt.ylabel("Package (LPA)")
plt.title("CGPA vs Package - Simple Linear Regression")
plt.legend()
plt.show()
```
